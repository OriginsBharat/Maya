```python
# Cell 5: Create API endpoint
import gradio as gr
from core.brain import Brain

brain = Brain()

def chat(message):
    return brain.think(message)

interface = gr.Interface(fn=chat, inputs="text", outputs="text")
interface.launch(share=True)  # Creates public URL
```
